```python
#1 задание
import requests
import json
from google.colab import files

# Введи GitHub username
username = input("Введите GitHub username: ").strip()

repos = []
page = 1

while True:
    url = f"https://api.github.com/users/{username}/repos"

    params = {
        "type": "public",
        "per_page": 100,
        "page": page
    }

    response = requests.get(url, params=params)

    if response.status_code == 404:
        print(f"Пользователь '{username}' не найден.")
        break

    response.raise_for_status()

    data = response.json()

    if not data:
        break

    repos.extend(data)
    page += 1


if repos:
    print(f"\публичных репозиториев: {len(repos)}\n")
    print("=" * 60)

    # Вывод информации в Google Colab
    for i, repo in enumerate(repos, 1):
        print(f"\n{i}. {repo['name']}")
        print(f"   Описание: {repo['description'] or 'Нет описания'}")
        print(f"   Язык: {repo['language'] or 'Не указан'}")
        print(f"   Звёзды: {repo['stargazers_count']}")
        print(f"   Forks: {repo['forks_count']}")
        print(f"   URL: {repo['html_url']}")
        print(f"   Создан: {repo['created_at']}")
        print(f"   Обновлён: {repo['updated_at']}")

    print("\n" + "=" * 60)
    filename = "github_repositories.json"

    with open(filename, "w", encoding="utf-8") as file:
        json.dump(repos, file, ensure_ascii=False, indent=4)
    files.download(filename)

else:
    print("У пользователя нет публичных репозиториев.")
```

```python
#2 задание - авторизация
import requests
import json


TOKEN = input("Введите GitHub Token: ").strip()

headers = {
    "Authorization": f"Bearer {TOKEN}",
    "Accept": "application/vnd.github+json"
}


url = "https://api.github.com/user"

response = requests.get(url, headers=headers)

print("HTTP статус:", response.status_code)
print("\nОтвет сервера:")

if response.ok:
    data = response.json()
    print(json.dumps(data, indent=4, ensure_ascii=False))
else:
    print(response.text)
```
